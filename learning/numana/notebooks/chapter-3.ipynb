## 3.2.2c

We will use the first implementation seen in the provided notebooks for this question.

```python
def q(xs,ys,i,j,t):
    if j == 0:
        return ys[i]
    elif j > i:
        return 0
    else:
        return ((t-xs[i-j])*q(xs,ys,i,j-1,t) - (t-xs[i])*q(xs,ys,i-1,j-1,t))/(xs[i]-xs[i-j])

# Note that this is calculating with the entire array
def approx(xs,ys,t):
    return q(xs,ys,len(xs)-1,len(ys)-1,t)

xs = [0.1, 0.2, 0.3, 0.4]
ys = [-0.29004986, -0.56079734, -0.81401972, -1.0526302]
t = 0.18

# So when we use approx(), we actually need to filter the arrays out ourselves
p1 = approx(xs[:2], ys[:2], t)
p2 = approx(xs[:3], ys[:3], t)
p3 = approx(xs[:4], ys[:4], t)

print(p1, p2, p3)
```

## 3.3.4b

The exercise requires us to (1) find the coefficients of the relevant polynomials, and (2) evaluate $P_n(0.18)$. 
Using the formula provided, we have:
$$
P_n(x) = f(x_0) + \sum_{k=1}^n \binom{s}{k}\Delta^k f(x_0),
\qquad s = \frac{x - x_0}{h}
$$

```python
# define a function that calculates the iterated differences
def forward_diffs(y):
    diffs = [np.array(y, dtype=float)]  # make an array/table
    while len(diffs[-1]) > 1:           # runs until we get delta^(number of points-1)
        diffs.append(np.diff(diffs[-1]))# finds the difference between current/previous val, appends it to array
    return diffs

# define another f that implements the forward difference formula
def newton_est(x, x0, h, diffs, degree):
    s = (x - x0) / h
    p = diffs[0][0]                     # p is initially just f(x0)
    term = 1.0                          # since s choose 0 is 1
    for k in range(1, degree + 1):
        term *= (s - (k - 1)) / k       # this is supposed to mimic (s choose k) for each k value
        p += term * diffs[k][0]         # each loop adds (s choose k) * delta^k f(x0)
    return p

x = [0.1, 0.2, 0.3, 0.4]
y = [-0.29004986, -0.56079734, -0.81401972, -1.0526302]

# first degree estimation
p41 = newton_est(0.18, x[0], x[1] - x[0], diffs, 1)
print("p1(0.18) =", p41)
print(diffs[0][0], diffs[1][0])
```

## 3.3.8a

We use the functions provided to calculate the coefficients. Since the table has 5 pairs of x-y values, a degree-4 polynomial is needed to interpolate them.

```python
# Declaring newton polynomial functions.
def newton_coeffs(xs,ys):
    n = len(xs)
    t = np.zeros((n,n))
    for i,y in enumerate(ys):
        t[i][0] = y
    for col in range(1,n):
        for row in range(col,n):
            t[row][col] = (t[row][col-1]-t[row-1][col-1])/(xs[row]-xs[row-col])
    return [t[i][i] for i in range(0,n)]
    
xs = np.array([0.0, 0.1, 0.3, 0.6, 1.0])
ys = np.array([-6.00000, -5.89483, -5.65014, -5.17788, -4.28172])
a4 = newton_coeffs(xs,ys)
print(a4)
```

## 3.4.2d

Using the provided functions, we implement the Hermite polynomial interpolation process below:

```python
def hermite_coeffs(xs,ys,yps):
    n = len(xs)
    t = np.zeros((2*n,2*n))
    for i in range(0,n):
        t[2*i][0] = ys[i]
        t[2*i+1][0] = ys[i]
    for i in range(1,n):
        t[2*i][1] = (t[2*i][0]-t[2*i-1][0])/(xs[i]-xs[i-1])
        t[2*i-1][1] = yps[i-1]
    t[2*n-1][1] = yps[n-1]
    for col in range(2,2*n):
        for row in range(col,2*n):
            t[row][col] = (t[row][col-1]-t[row-1][col-1])/(xs[row//2]-xs[(row-col)//2])
    return [t[i][i] for i in range(0,2*n)]

xs = np.array([-1, -0.5, 0, 0.5])
ys = np.array([0.86199480, 0.95802009, 1.0986123, 1.2943767])
yps = np.array([0.15536240, 0.23269654, 0.33333333, 0.45186776])

a5 = hermite_coeffs(xs,ys,yps)
print(a5)
print(len(a5))
```

## 3.4.4d

Thankfully, we do not need to write the Hermite polynomial's coefficients out every time to calculate our expected values. Using the function below, we can calculate $H(0.25)$, and find the absolute error between the data generating function $f(x) = \ln(e^x + 2)$: 

```python
def make_hermite_poly(xs,ys,yps):
    c = hermite_coeffs(xs,ys,yps)
    n = len(c)
    def f(t):
        sum = 0
        for i in range(0,n):
            prod = c[i]
            for j in range(0,i):
                prod = prod*(t-xs[j//2])
            sum = sum + prod
        return sum
    return f

ph = make_hermite_poly(xs,ys,yps)

# defining the target function
def f(x):
    return np.log(np.exp(x)+2)

print("Absolute error is approximately", np.abs(f(0.25)-ph(0.25)))
```

As we can see, the Hermite polynomial seems to track the target function quite effectively.
