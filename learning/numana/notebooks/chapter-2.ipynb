## 2.2.14a

We first can convert this quite easily into another fixed point problem: $g(x) = 2+\sin(x) = x$. 
Then, we'd need to verify that $g(x)$ has a fixed point on $[2,3]$. Since $0<\sin(x)<1$ when $x \in [2,3]$, we can be sure that the function remains in the defined interval.

With this in mind, we try to find an upper bound for the contraction constant of this recursion. Note that

$$
g'(x) = 0 + \sin'(x) = \cos(x)
$$

which is negative and monotonically decreasing on the interval [2,3]. This means that $|\cos(x)|$ is increasing, and the contraction constant 

$$
k = \max_{x \in [2,3]} |g'(x)|= |\cos(3)| \approx 0.9899
$$

Since we don't have a specified initial value, we'll try $x_0 = 2$. Using the recursion, $x_1 \approx 2.9093$.

Plugging the values into the formula for the error estimate nets us:

$$
|p-x_n| \leq \frac{k^n}{1-k} |x_1-x_0| \\

\frac{0.9899^n}{1-0.9899} \times (0.9093) \geq |p-x_n|
$$

Forcing this upper bound to be less than $10^{-5}$ nets us a result of $n \gtrsim 1592$. 

We finally try to implement this in python, using the provided script:

```python
def h(x):
    return 2 + np.sin(x)

ls5 = iter(h, 2, 63)
print(ls5)

# simple logic test to make sure the difference between the iterations are actually smaller than 10^-5
np.float64(10**-5) > np.abs((h(ls5[-1])-ls5[-1]))

# 63 is the first value of n such that |g(p_n) - p_n| is smaller than 10^-5
```

From actually running the calculations, we now know that it only takes $\approx$ 64 iterations (compared to the estimated 1593 iterations) for the difference between our iterations to be smaller than $10^{-5}$. This is expected, knowing that the results used in the derivations above represent the absolute worst cases for our process.

## 2.3.6b + 2.3.8b
We first calculate the derivative of the function $f(x) = \ln(x-1) + \cos(x-1)$:
$$
f'(x) = \frac{1}{x-1} - \sin(x-1)
$$
Then we implement Newton's method in Python with the provided scripts, using an initial value of 1.6:

```python
def newton(fn,fnd,st,n):
    def gn(x):
        return x - fn(x)/fnd(x)
    ls = iter(gn,st,n)
    return ls

def fb(x):
    return np.log(x - 1) + np.cos(x - 1)

def fbd(x):
    return 1/(x - 1) - np.sin(x - 1)

x = newton(fb,fbd,1.6,8)
print(x)
```

We can see that about 4 iterations in, new iterations share at least 4 decimal places with the old ones, indicating that we are likely within $10^{-4}$ of our solution. With this in mind, we can move on to the secant method, whose implementation is shown below:

```python
def secant(fn,pr):
    p0 = pr[0]
    p1 = pr[1]
    p2 = p1 - fn(p1)/((fn(p1)-fn(p0))/(p1-p0))
    return (p1,p2)
def secant_iter(fn,pr,n):
    prs = []
    p = pr
    for i in range(0,n):
        prs.append(p)
        p = secant(fn,p)
    return prs

# For demonstrative purposes, we can use the bounds provided as the first pair for this method
a = secant_iter(fb,(1.3,2.0),10)
print(a)
```

We see that with the secant method, it took a few more iterations before the method produced estimates close to the final iterations in Newton's method. However, this is expected, given Newton's method does require a closed-form formula for the derivative of the function. Regardless, we can be pretty sure that the estimates are within $10^{-4}$ of the true solution.

We are assuming monthly interest rates in this exercise, instead of an annual rate. With this in mind, we can do the first few calculations to make the problem a little more tractable. 20 years is 240 months.

$$
750000 = \frac{P}{i} \times ((1+i)^{240} - 1)
$$

Since the account holder can contribute $1500 a month, we can substitute P into the equation and arrive at:

$$
(1+i)^{240} = 500i + 1, \text{or} \\
(1+i)^{240} - 500i - 1 = 0
$$

The rest of this problem can only be solved numerically. Let $g(i) = (1+i)^{240} - 500i - 1$, we can find the derivative of the function:
$$
g'(i) = 240 \times (1+i)^{239} - 500
$$

This enables us to use Newton's method to find the roots of this function, whose implementation is shown below:

```python
def g(x):
    return np.power(1 + x, 240) - 500 * x - 1

def gd(x):
    return 240 * np.power(1 + x, 239) - 500

# having established the functions, we can wrap up with the newton iteration call
# we'll use 0.01 as our initial guess, though a 1% monthly interest rate does seem a little too good to be true

results = newton(g,gd,0.01,8)
print(results)
```

As shown by our results, the minimum interest rate to meet the $750000 goal is around 0.5551% every month.
