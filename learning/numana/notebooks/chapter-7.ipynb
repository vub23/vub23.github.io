## 7.3.6c & 7.3.8c

We are asked to solve a linear system using the provided algorithms. Since the implementations provided in the notebooks do not list tolerance, we would need to add a few details to both the Gauss Seidel method and the Jacobi method.

```python
# first, declare the matrix/vector
M2 = np.array([[4,1,-1,1],[1,4,-1,-1],[-1,-1,5,1],[1,-1,1,3]])
b2 = np.array([-2,-1,0,1])
st2 = np.array([0,0,0,0])

# then, declare the helpers
def upper(A):
    n = len(A)
    U = np.zeros((n,n))
    for i in range(0,n):
        for j in range(i+1,n):
            U[i][j] = -A[i][j]
    return U

def lower(A):
    n = len(A)
    L = np.zeros((n,n))
    for i in range(0,n):
        for j in range(0,i):
            L[i][j] = -A[i][j]
    return L

def diag(A):
    n = len(A)
    D = np.zeros((n,n))
    for i in range(0,n):
        D[i][i] = A[i][i]
    return D

# the implementation with tolerance
def jacobi_iter(A,b,st,n,TOL):
    t = np.dot(la.inv(diag(A)),lower(A)+upper(A))
    c = np.dot(la.inv(diag(A)),b)
    vs = [st]
    v = st
    for i in range(0,n):
        v_old = v
        v = np.dot(t,v)+c
        vs.append(v)
        if la.norm(v - v_old, ord=np.inf) < TOL:
            return vs
    return 'Max number of iterations reached'

def gauss_seidel_iter(A,b,st,n, TOL):
    t = np.dot(la.inv(diag(A)-lower(A)),upper(A))
    c = np.dot(la.inv(diag(A)-lower(A)),b)
    vs = [st]
    v = st
    for i in range(0,n):
        v_old = v
        v = np.dot(t,v)+c
        vs.append(v)
        if la.norm(v - v_old, ord=np.inf) < TOL:
            return vs
    return 'Max number of iterations reached'

# now, solve the equations
J2 = jacobi_iter(M2,b2,st2,20,10**-3)
GS2 = gauss_seidel_iter(M2,b2,st2,10,10**-3)
print(J2, '\n', GS2)
```

From our experimentation, the Jacobi method did take more iterations to reach the required tolerance versus the Gauss-Seidel method.


## 7.4.6c
We are once again asked to solve a linear system, but this time using the SOR method. The provided implementation does not make use of tolerances, so appropriate changes will be added.

```python
def sor(A,b,omega,st,n,TOL):
    t = np.dot(la.inv(diag(A)-omega*lower(A)),((1-omega)*diag(A)+omega*upper(A)))
    c = omega*np.dot(la.inv(diag(A)-omega*lower(A)),b)
    vs = [st]
    v = st
    for i in range(0,n):
        v_old = v
        v = np.dot(t,v)+c
        vs.append(v)
        if la.norm(v - v_old, ord=np.inf) < TOL:
            return vs
    return 'Max number of iterations reached'

M3 = np.array([[4,1,-1,1],[1,4,-1,-1],[-1,-1,5,1],[1,-1,1,3]])
b3 = np.array([-2,-1,0,1])
st3 = np.array([0,0,0,0])

sol3 = sor(M3,b3,1.2,st3,6,10**-3)
print(sol3)
```

## 7.6.6d
We are asked to implement the preconditioned conjugate gradient method to solve the same system of linear equations, with a provided preconditioning matrix.
For the implementation below, I assume that we are still working under the infinity norm (like the previous questions using the same system but requesting solutions through different methods). The number of iterations is also reduced to 2. Given the small number of iterations, the pre-set tolerance being the stopping condition is unlikely.


```python
M5 = np.array([[4,1,-1,1],[1,4,-1,-1],[-1,-1,5,1],[1,-1,1,3]])
b5 = np.array([-2,-1,0,1])
st5 = np.array([0,0,0,0])

# the grad method implementation
def precon_grad(A, b, st, n, TOL):
    # since D^-1/2 is just a diagonal array, we might not need to use the entire matrix
    D = np.diag(A)
    Dsqrt = 1/np.sqrt(D)
    x = st
    r = b - np.dot(A, x)
    w = Dsqrt * r
    v = Dsqrt * w # since transpose of diag is the same
    alpha = np.dot(w,w)
    for _ in range(n):
        if la.norm(v, ord = np.inf) < TOL:
            return 'solution vector', x, 'with residual', r
        u = np.dot(A,v)
        t = alpha/np.dot(v,u)
        x = x + t*v
        r = r - t*u
        w = Dsqrt * r
        beta = np.dot(w,w)
        if np.abs(beta) < TOL:
            if la.norm(r, ord = np.inf) < TOL:
                return "solution vector", x, "with residual", r
        s = beta/alpha
        v = Dsqrt*w + s*v
        alpha = beta
    return "solution vector", x, "with residual", r

sol5 = precon_grad(M5, b5, st5, 2, 10**-3)
print(sol5)
```
