## 5.2.6c + 5.2.8c
We are to solve an initial-value problem where:
$$
y' = \frac{1}{t} \times (y^2+y),
$$
with $1 \leq t \leq 3$, h = 0.2 and y(1) = -2. We have an Euler implementation on hand, and will use that to generate our estimates.

```python
def euler(f,t0,y0,h,n):
    y = y0
    t = t0
    pts = [(t,y)]
    for _ in range(0,n):
        y = y + h * f(t,y)
        t = t + h
        pts.append((t,y))
    return pts

def f1(t,y):
    return 1/t*(y**2+y)

ea1 = euler(f1, 1, -2, 0.2, 10)
print(ea1)
```

Having arrived at our solutions, we can now use the actual function to calculate the values and print the errors:

```python
def soln1(t):
    return 2*t/(1-2*t)

t1 = np.linspace(1,3,11)
y1 = soln1(t1)
est1 = np.array([pt[1] for pt in ea1])
err1 = list(zip(t1, y1 - est1))
print(err1)
```

## 5.3.6c
From our derivations (taking the derivative of y' with respect to t, keeping in mind $(y^2)' = 2yy'$, we can see that:
$$
y'' = \frac{2(y^3+y^2)}{t^2}
$$.
Updating our initial Euler implementation, we can calculate our estimates:

```python
def taylor2(f,f2,t0,y0,h,n):
    y = y0
    t = t0
    pts = [(t,y)]
    for _ in range(0,n):
        y = y + h * f(t,y) + h**2/2 * f2(t,y)
        t = t + h
        pts.append((t,y))
    return pts

def f12(t,y):
    return (2*y**3 + 2*y**2)/t**2

ea2 = taylor2(f1,f12, 1, -2, 0.2, 10)
print(ea2)
```

## 5.4.4b
We are asked to use the modified Euler method to approximate the solutions to the equation
$$
y' = \frac{y^2}{1+t}
$$
and to compare our estimates with the actual solution.

```python
def mod_euler(f,t0,y0,h,n):
    y = y0
    t = t0
    pts = [(t,y)]
    for _ in range(0,n):
        y = y + h/2 * (f(t,y) + f(t+h, y + h*f(t,y)))
        t = t + h
        pts.append((t,y))
    return pts

def f2(t,y):
    return y**2/(1+t)

def soln2(t):
    return -1 / np.log(t+1)

ea3 = mod_euler(f2, 1, -1/np.log(2), 0.1, 10)

esoln = soln2(2)
err3 = ea3[-1][1] - esoln
print("At the last step, the error is", err3)
```

## 5.4.16b

With the same initial value problem, we will estimate the solution using the Runge-Kutta method. The implementation will be done below:

```python
def runge_kutta(f,t0,y0,h,n):
    y = y0
    t = t0
    pts = [(t,y)]
    for _ in range(0,n):
        k1 = f(t,y)
        k2 = f(t+h/2,y+k1*h/2)
        k3 = f(t+h/2,y+k2*h/2)
        k4 = f(t+h,y+k3*h)
        y = y + (k1+2*k2+2*k3+k4)*h/6
        t = t + h
        pts.append((t,y))
    return pts

ea4 = runge_kutta(f1, 1, -2, 0.2, 10)
print("Estimation array:", ea4)
est4 = np.array([pt[1] for pt in ea4])
err4 = list(zip(t1, y1 - est4))
print("Error array:", err4)
```

As we can see, the RK method is much more accurate compared to the other methods.

## 5.5.2d
We are to estimate the solution to another problem using the Runge-Kutta-Fehlberg method:
$$
y' = t^2
$$
with a tolerance of $10^{-4}$ and other parameters. Since an implementation of RKF in python, the solution can be estimated as seen below:

```python
# changing tolerance to desired amount 
def rkf(f,t0,y0,t1, hmin, hmax):
    TOL = 10**(-4)
    def step(t,h,y):
        k1 = h*f(t,y)
        k2 = h*f(t + h/4, y + k1/4)
        k3 = h*f(t + 3*h/8, y + 3*k1/32 + 9*k2/32)
        k4 = h*f(t + 12*h/13,y + 1932*k1/2197 - 7200*k2/2197 + 7296*k3/2197)
        k5 = h*f(t + h, y + 439*k1/216 - 8*k2 + 3680*k3/513 - 845*k4/4104)
        k6 = h*f(t + h/2,y - 8*k1/27 + 2*k2 - 3544*k3/2565 + 1859*k4/4104 - 11*k5/40)
        r = abs(k1/360 - 128*k3/4275 - 2197*k4/75240 + k5/50 + 2*k6/55)/h
        q = 0.84*(TOL/r)**(1/4)
        if q <= 0.1:
            newH = 0.1*h
        elif q >= 4:
            newH = 4*h
        else:
            newH = q*h
        if newH > hmax:
            newH = hmax
        if newH < hmin: 
            raise Exception("Minimum step size exceeded")
        if r >= TOL:
            return step(t,newH,y)
        else:
            return (t+h,newH,y + 25*k1/216 + 1408*k3/2565 + 2197*k4/4104 - k5/5)
    pt = (t0,hmax,y0) # making sure 1st step uses hmax
    pts = [pt]
    while pt[0] < t1:
        h = min(pt[1], t1 - pt[0]) # making sure that the step size is clipped so that it doesn't go beyond the bounds
        pt = step(pt[0],h,pt[2])
        pts.append(pt)
    return pts

# declaring the relation
def f3(t,y):
    return t**2

# setting a starting value equal to the max allowed
ea5 = rkf(f3, 0, 0, 2, 0.02, 0.5)
print(ea5)
```

Since the solution to this problem is $y = t^3/3$, it is not difficult to see that the algorithm quickly converged to the right estimates.
